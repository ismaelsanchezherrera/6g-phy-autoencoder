```python
import torch
import numpy as np
import matplotlib.pyplot as plt

print(f"Versión de PyTorch: {torch.__version__}")
print(f"Soporte CUDA disponible: {torch.cuda.is_available()}")

if torch.cuda.is_available():
    print(f"Dispositivo GPU activo: {torch.cuda.get_device_name(0)}")
else:
    print("Atención: Ejecutando en CPU.")
```

```python
# M = 16 simbolos (equivalente a 16-QAM en capacidad de representacion)
M = 16
batch_size = 5

# 1. generar mensajes aleatorios
labels = torch.randint(0, M, (batch_size,))

# 2. convertir a representación One-Hot
one_hot = torch.nn.functional.one_hot(labels, num_classes=M).float()

# 3. simulacion de ruido AWGN sintetico (std_dev = 0.5)
noise = torch.randn_like(one_hot) * 0.5
received_signal = one_hot + noise

print("Etiquetas originales:", labels.numpy())
print("Forma del vector One-Hot:", one_hot.shape)
print("Forma de la señal recibida:", received_signal.shape)
```
